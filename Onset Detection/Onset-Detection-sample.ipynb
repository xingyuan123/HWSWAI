```python
# sample pseudo code for Onset-detection
def Onset(x_acc, y_acc, z_acc, x_gyro, y_gyro, z_gyro):
    threshold = 3000
    x = 5

    oldest = x_acc[0] + y_acc[0] + z_acc[0] + x_gyro[0] + y_gyro[0] + z_gyro[0]
    newest_x_minus_1 = 0
    for i in range(1, x):
        newest_x_minus_1 += x_acc[i] + y_acc[i] + z_acc[i] + x_gyro[i] + y_gyro[i] + z_gyro[i]
    newest_x_minus_1 /= (x - 1)

    if newest_x_minus_1 - oldest > threshold:
        return True
    return False
```
